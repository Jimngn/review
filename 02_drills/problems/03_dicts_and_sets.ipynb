## 1. Most common element

Write `most_common(items)` that returns the element that appears most often in the list. Assume no ties.
```
most_common(["a", "b", "a", "c", "a"]) -> "a"
```

````python
def most_common(items):
    count = {}
    for i in items:
        count[i] = count.get(i, 0) + 1
    high = 0
    res = None
    for item, cnt in count.items():
        if cnt > high:
            high = cnt
            res = item
    return res
````

````python
def most_common(items):
    count = {}
    for i in items:
        count[i] = count.get(i, 0) + 1
    return max(count, key=count.get)
````

````python
most_common(["a", "b", "a", "c", "a"]) 
````

````python
assert most_common(["a", "b", "a", "c", "a"]) == "a"
print("looks good")
````

## 2. Anagram check

Write `is_anagram(s1, s2)` that returns `True` if `s1` and `s2` contain exactly the same letters in the same quantities (ignore case), `False` otherwise.
```
is_anagram("listen", "silent") -> True
is_anagram("hello", "world")   -> False
```

````python
def is_anagram(s1, s2):
    count_s1 = {}
    count_s2 = {}
    for i in s1:
        count_s1[i] = count_s1.get(i, 0) + 1
    for i in s2:
        count_s2[i] = count_s2.get(i, 0) + 1
    return count_s1 == count_s2
    
        
````

````python
assert is_anagram("listen", "silent") == True
assert is_anagram("hello", "world") == False
print("looks good")
````

## 3. Group words by length

Write `group_by_length(words)` that returns a dict mapping each length to a list of words of that length, in original order.
```
group_by_length(["a", "bb", "cc", "ddd"]) -> {1: ["a"], 2: ["bb", "cc"], 3: ["ddd"]}
```

````python
def group_by_length(words):
    group = {}
    for i in words:
        n = len(i)
        if n not in group:
            group[n] = [i]
        else:
            group[n].append(i)
    return group
````

````python
assert group_by_length(["a", "bb", "cc", "ddd"]) == {1: ["a"], 2: ["bb", "cc"], 3: ["ddd"]}
print("looks good")
````

## 4. Two-sum

Write `two_sum(nums, target)` that returns a tuple `(i, j)` of two indices whose values add up to `target`, or `None` if no such pair exists. Assume at most one valid pair.
```
two_sum([2, 7, 11, 15], 9) -> (0, 1)     # 2 + 7 == 9
two_sum([1, 2, 3], 100)    -> None
```
(This is a classic — try to do it in one pass with a dict of "value seen so far -> index", rather than a nested loop checking every pair.)

````python
def two_sum(nums, target):
    seen = {}
    for i in range(len(nums)):
        complement = target - nums[i]
        if complement in seen:
            return (seen[complement], i)
        else:
            seen[nums[i]] = i
````

````python
print(two_sum([2, 7, 11, 15], 9))
````

````python
assert two_sum([2, 7, 11, 15], 9) == (0, 1)
assert two_sum([1, 2, 3], 100) is None
print("looks good")
````

## 5. Find all duplicates

Write `find_duplicates(items)` that returns a set of all values that appear more than once.
```
find_duplicates([1, 2, 3, 2, 4, 1]) -> {1, 2}
```

````python
def find_duplicates(items):
    dup = set()
    seen = set()
    for i in items:
        if i not in seen:
            seen.add(i)
        else:
            dup.add(i)
    return dup
````

````python
assert find_duplicates([1, 2, 3, 2, 4, 1]) == {1, 2}
print("looks good")
````

## 6. Invert a dictionary

Write `invert(d)` that returns a new dict with keys and values swapped. Assume all values in `d` are unique (so the inversion is well-defined).
```
invert({"a": 1, "b": 2}) -> {1: "a", 2: "b"}
```

````python
def invert(d):
    new = {}
    for key, values in d.items():
        new[values] = key
    return new
````

````python
invert({"a": 1, "b": 2})
````

````python
assert invert({"a": 1, "b": 2}) == {1: "a", 2: "b"}
print("looks good")
````
