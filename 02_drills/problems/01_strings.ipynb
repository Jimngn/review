## 1. Reverse a string

Write `reverse(s)` that returns the string reversed.
```
reverse("hello") -> "olleh"
```

````python
def reverse(s):
    return s[::-1]
````

````python
assert reverse("hello") == "olleh"
assert reverse("") == ""
print("looks good")
````

## 2. Is palindrome

Write `is_palindrome(s)` that returns `True` if `s` reads the same forwards and backwards, `False` otherwise. Treat it as case-sensitive for now (no need to handle spaces/punctuation).
```
is_palindrome("racecar") -> True
is_palindrome("hello")   -> False
```

````python
def is_palindrome(s):
    return s == s[::-1]
````

````python
assert is_palindrome("racecar") == True
assert is_palindrome("hello") == False
print("looks good")
````

## 3. Count vowels

Write `count_vowels(s)` that returns the number of vowels (a, e, i, o, u — either case) in `s`.
```
count_vowels("Hello World") -> 3
```

````python
def count_vowels(s):
    vowels = set(("a", "e", "i", "o", "u"))
    count = 0
    for ch in s:
        if ch in vowels:
            count += 1
    return count
    
````

````python
assert count_vowels("Hello World") == 3
assert count_vowels("xyz") == 0
print("looks good")
````

## 4. Word count

Write `word_count(s)` that returns the number of words in `s`, where words are separated by whitespace (there may be multiple spaces between words, or leading/trailing spaces).
```
word_count("  the quick  brown fox ") -> 4
```

````python
def word_count(s):
    count = 0
    for i in range(len(s) - 1):
        if s[i] == " " and s[i+1] != " ":
            count += 1
    return count
            
            
        
````

````python
assert word_count("  the quick  brown fox ") == 4
assert word_count("") == 0
print("looks good")
````

## 6. Caesar cipher

Write `caesar_shift(s, n)` that shifts every lowercase letter in `s` forward by `n` positions in the alphabet, wrapping around from 'z' back to 'a'. You can assume `s` is all lowercase letters and spaces, and spaces are left unchanged. (Hint: `ord()` gives you a character's numeric code, `chr()` goes the other way.)
```
caesar_shift("abc xyz", 2) -> "cde zab"
```

````python
def caesar_shift(s, n):
    new_string = ""
    for ch in s:
        if ch == " ":
            new_string += " "
        else:
            if ord(ch)+n > 122:
                new_string += chr(ord(ch) + n - 26)
            else:
                new_string += chr(ord(ch) + n)

    return new_string
````

````python
caesar_shift("abc xyz", 2)
````

````python
assert caesar_shift("abc xyz", 2) == "cde zab"
print("looks good")
````
