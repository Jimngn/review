## 1. FizzBuzz

Write `fizzbuzz(n)` that returns a list of strings for numbers 1 through `n`: `"Fizz"` for multiples of 3, `"Buzz"` for multiples of 5, `"FizzBuzz"` for multiples of both, otherwise the number itself as a string.
```
fizzbuzz(15) -> ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
```

````python
def fizzbuzz(n):
    result = []
    for i in range(1,n+1):
        if i % 3 == 0 and i % 5 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result
````

````python
print(fizzbuzz(15))
````

````python
assert fizzbuzz(15) == ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]
print("looks good")
````

## 2. Is prime

Write `is_prime(n)` that returns whether `n` is a prime number. Assume `n` is a positive integer.
```
is_prime(7) -> True
is_prime(8) -> False
is_prime(1) -> False
```

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True
````

````python
assert is_prime(7) == True
assert is_prime(8) == False
assert is_prime(1) == False
print("looks good")
````

## 3. Fibonacci

Write `fibonacci(n)` that returns the `n`th Fibonacci number (0-indexed: `fibonacci(0) == 0`, `fibonacci(1) == 1`). Do it with a loop, not recursion (recursion works but is much slower for this without extra machinery).
```
fibonacci(0) -> 0
fibonacci(1) -> 1
fibonacci(6) -> 8
```

````python
def fibonacci(n):
    a, b = 0, 1
    for i in range(n):
        a, b = a+b, a
    return a
````

````python
assert fibonacci(0) == 0
assert fibonacci(1) == 1
assert fibonacci(6) == 8
print("looks good")
````

## 4. Validate a simple password

Write `is_valid_password(s)` that returns `True` if `s` is at least 8 characters long, contains at least one digit, and contains at least one uppercase letter. Otherwise `False`.
```
is_valid_password("Abcdefg1") -> True
is_valid_password("abcdefg1") -> False   # no uppercase
is_valid_password("Abc1")     -> False   # too short
```

````python
def is_valid_password(s):
    if len(s) < 8:
        return False
    a =  any(char.isdigit() for char in s)
    b = any(char.isupper() for char in s)
    return a and b
````

````python
assert is_valid_password("Abcdefg1") == True
assert is_valid_password("abcdefg1") == False
assert is_valid_password("Abc1") == False
print("looks good")
````

## 5. Greatest common divisor

Write `gcd(a, b)` that returns the greatest common divisor of two positive integers, using Euclid's algorithm (`gcd(a, b) == gcd(b, a % b)`, with `gcd(a, 0) == a`).
```
gcd(48, 18) -> 6
```

````python
def gcd(a, b):
    while b != 0:
        a, b = b, a%b
    return a
````

````python
assert gcd(48, 18) == 6
print("looks good")
````

## 6. Apply a function n times

Write `repeat(f, x, n)` that applies function `f` to `x`, `n` times in a row, and returns the final result.
```
repeat(lambda x: x * 2, 3, 4) -> 48    # 3 -> 6 -> 12 -> 24 -> 48
```

````python
def repeat(f, x, n):
    for i in range(n):
        x = f(x)
    return x
        
````

````python
assert repeat(lambda x: x * 2, 3, 4) == 48
print("looks good")
````
