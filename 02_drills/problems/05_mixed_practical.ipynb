## 1. Top three trade sizes

Given a list of trade sizes (ints), return the three largest, in descending order. If there are fewer than three trades, return as many as there are, still sorted descending.
```
top_three([40, 10, 90, 30, 90, 5]) -> [90, 90, 40]
top_three([5, 1])                  -> [5, 1]
top_three([])                      -> []
```

````python
def top_three(trades):
    trades.sort(reverse = True)
    return trades[:3]
````

````python
assert top_three([40, 10, 90, 30, 90, 5]) == [90, 90, 40]
assert top_three([5, 1]) == [5, 1]
assert top_three([]) == []
print("looks good")
````

## 2. Largest total by symbol

Given a list of `(symbol, quantity)` tuples representing trades, return the symbol with the largest **total** quantity traded (summed across all its trades). Assume no ties.
```
largest_total([("AAPL", 100), ("GOOG", 50), ("AAPL", 80)]) -> "AAPL"   # AAPL total = 180
```

````python
def largest_total(trades):
    book = {}
    for sym, qty in trades:
        if sym not in book:
            book[sym] = qty
        else:
            book[sym] += qty
            #only use enumerate for index and then value not key and value
    return max(book, key = book.get)
````

````python
largest_total([("AAPL", 100), ("GOOG", 50), ("AAPL", 80)])
````

````python
assert largest_total([("AAPL", 100), ("GOOG", 50), ("AAPL", 80)]) == "AAPL"
print("looks good")
````

## 3. Parse and total a trade log

You're given a multi-line string, one trade per line, formatted `"SYMBOL,PRICE,QUANTITY"`. Write `total_value(log)` that returns the total dollar value of all trades (`price * quantity`, summed).
```python
log = """AAPL,150.0,10
GOOG,2800.0,2
AAPL,151.5,5"""
total_value(log) -> 7857.5     # 150*10 + 2800*2 + 151.5*5
```

````python
def total_value(log):
    total = 0
    for line in log.strip().split("\n"):
        sym, price, qty = line.split(",")
        total += float(qty) * float(price)
    return total

#strip removes leading/trailing whitespaces
#split splits by character/
````

````python
log = """AAPL,150.0,10
GOOG,2800.0,2
AAPL,151.5,5"""
print(total_value(log))
````

````python
log = """AAPL,150.0,10
GOOG,2800.0,2
AAPL,151.5,5"""
assert total_value(log) == 7857.5
print("looks good")
````

## 4. Running maximum

Given a list of numbers, return a new list where each position holds the maximum value seen so far (including the current position).
```
running_max([3, 1, 4, 1, 5, 9, 2]) -> [3, 3, 4, 4, 5, 9, 9]
```

````python
def running_max(nums):
    result = []
    curr = 0
    for num in nums:
        curr = max(curr, num)
        result.append(curr)
    return result
````

````python
assert running_max([3, 1, 4, 1, 5, 9, 2]) == [3, 3, 4, 4, 5, 9, 9]
print("looks good")
````

## 5. Matching buy/sell orders

You have two lists of equal length: `buys` and `sells`, where `buys[i]` and `sells[i]` are the price a trader is willing to buy/sell the `i`th unit at. A trade executes at position `i` if `buys[i] >= sells[i]`. Write `count_matches(buys, sells)` returning how many positions would execute.
```
count_matches([10, 5, 8], [9, 6, 8]) -> 2    # positions 0 and 2 match
```

````python
def count_matches(buys, sells):
    n = len(buys)
    result = 0
    for i in range(n):
        if buys[i] >= sells[i]:
            result += 1
    return result
````

````python
assert count_matches([10, 5, 8], [9, 6, 8]) == 2
print("looks good")
````

## 6. Generalize: top k

Extend problem 1 to a general `top_k(nums, k)`. Don't sort the *entire* list if you can avoid it when `k` is small relative to `len(nums)` — but a fully correct `sorted(nums, reverse=True)[:k]` is a perfectly good first answer; only optimize further if asked.
```
top_k([40, 10, 90, 30, 90, 5], 2) -> [90, 90]
top_k([7], 5) -> [7]
```
Follow-up to think through (no code needed, just be ready to answer out loud): what would you do differently if trades arrived **one at a time** (a stream) instead of as a full list up front, and you needed to always be able to report the current top k?

````python
def top_k(nums, k):
    nums.sort(reverse = True)
    return nums[:k]
````

````python
assert top_k([40, 10, 90, 30, 90, 5], 2) == [90, 90]
assert top_k([7], 5) == [7]
print("looks good")
````
